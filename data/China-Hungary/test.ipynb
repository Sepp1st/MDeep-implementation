```python
import numpy as np
import pandas as pd
```

```python
#load the data
# Update the path below to the correct location of your .npz file
data = np.load('China_Hung_abundance.npz', allow_pickle=True)
```

```python
#check the keys
data.keys()
```

```python
#check sample ids
sample_ids = data['sample_ids']
print("Sample IDs:", sample_ids)
```

```python
Y = np.array([1 if 'BC' in sample_id or 'U' in sample_id else 0 for sample_id in sample_ids])
# print the Y array
print("Y array:", Y)
# zip and print side by side
for sample_id, y_value in zip(sample_ids, Y):
    print(f"Sample ID: {sample_id}, Y: {y_value}")
# export Y to .npy
np.save('Y.npy', Y)
```

```python
X = np.load('X.npy', allow_pickle=True)
Y = np.load('Y.npy', allow_pickle=True)
from sklearn.model_selection import train_test_split
# Split the data into training and testing sets
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
# Print the shapes of the training and testing sets
print("X_train shape:", X_train.shape)
print("Y_train shape:", Y_train.shape)
print("X_test shape:", X_test.shape)
print("Y_test shape:", Y_test.shape)
# Save the split data as .npy files
np.save('X_train.npy', X_train)
np.save('Y_train.npy', Y_train)
np.save('X_eval.npy', X_test)
np.save('Y_eval.npy', Y_test)
```
