```python
import numpy as np
import pandas as pd
```

```python
#load the data
# Update the path below to the correct location of your .npz file
data = np.load('3_countries_abundance.npz', allow_pickle=True)
```

```python
#check the keys
data.keys()
#check size of the data
print("Data shape:", data['matrix'][0])
```

```python
#check sample ids
sample_ids = data['sample_ids']
print("Sample IDs:", sample_ids)
```

```python
# create corresponding Y
# if sample_ids contains ["BC","U"] then Y should be 1
# else Y should be 0
Y = np.array([1 if 'BC' in sample_id or 'U' in sample_id else 0 for sample_id in sample_ids])
# print the Y array
print("Y array:", Y)
# zip and print side by side
for sample_id, y_value in zip(sample_ids, Y):
    print(f"Sample ID: {sample_id}, Y: {y_value}")
# export Y to .npy
#np.save('Y.npy', Y)
```
